```python

if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

df_clean = df.dropna().drop_duplicates()


total_tracks = len(df_clean)
unique_artists = df_clean['artists'].nunique()
unique_genres = df_clean['track_genre'].nunique()

pop_75 = df_clean[df_clean['popularity'] > 75]
pop_0 = df_clean[df_clean['popularity'] == 0]

print(f"Total Clean Tracks: {total_tracks}")
print(f"Unique Artists: {unique_artists}")
print(f"Unique Genres: {unique_genres}")
print(f"Tracks with Popularity > 75: {len(pop_75)} ({len(pop_75)/total_tracks*100:.2f}%)")
print(f"Tracks with Popularity = 0: {len(pop_0)} ({len(pop_0)/total_tracks*100:.2f}%)")

print("\n--- Audio Profile Comparison ---")
comp = pd.DataFrame({
    'Overall Mean': df_clean[['danceability', 'energy', 'loudness', 'acousticness', 'valence', 'tempo']].mean(),
    'Popular Tracks (>75) Mean': pop_75[['danceability', 'energy', 'loudness', 'acousticness', 'valence', 'tempo']].mean()
})
print(comp.round(3))
```

```python
numeric_features = [
    'danceability', 'energy', 'loudness', 'speechiness',
    'acousticness', 'instrumentalness', 'liveness', 'valence',
    'tempo', 'duration_ms', 'popularity'
]

for col in numeric_features:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

df = df.dropna(subset=['popularity'])

df['duration_min'] = df['duration_ms'] / 60000


```

```python
print("Top 5 artits")
print(df['artists'].value_counts().head(5))
print("TOP 5 ARTISTS BY AVERAGE POPULARITY")
top_artists_pop = df.groupby('artists')['popularity'].mean().nlargest(5)
print(top_artists_pop)
```

```python
artist_counts = df_clean['artists'].value_counts()
artists_gt_10 = artist_counts[artist_counts > 10]

print(artists_gt_10)
```

```python
artist_listening_min = (df_clean.groupby('artists')['duration_ms'].sum() / 60000).round(2)
top_listening_artists = artist_listening_min.sort_values(ascending=False)
print(top_listening_artists.head(10))

valid_artists = df_clean['artists'].value_counts()[lambda x: x > 10].index
print(valid_artists)
top_by_time = (df_clean[df_clean['artists'].isin(valid_artists)].groupby('artists')['duration_ms'].sum() / 60000).sort_values(ascending=False)
print(top_by_time)
```

```python
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])
df_clean = df.dropna().drop_duplicates()
valid_artists = df_clean['artists'].value_counts()[lambda x: x > 10].index
artist_profiles = df_clean[df_clean['artists'].isin(valid_artists)].groupby('artists').agg(
    song_count=('track_id', 'count'),
    mean_popularity=('popularity', 'mean'),
    danceability=('danceability', 'mean'),
    energy=('energy', 'mean'),
    valence=('valence', 'mean'),
    loudness=('loudness', 'mean'),
    tempo=('tempo', 'mean')
).round(3)


ranked_artists = artist_profiles.sort_values(by='mean_popularity', ascending=False)
print(ranked_artists.head(20))
```
